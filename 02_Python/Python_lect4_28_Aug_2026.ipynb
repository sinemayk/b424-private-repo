```python
# öğrenciden aldığınız not eğer 70 ve üzeri ise "A ile geçtiniz"
# 50-69 arası ise "B ile geçtiniz"
# 50'nin altındaysa "Kaldınız" yazdırın.
```

```python
note = int(input("Lütfen notunuzu giriniz: "))
if note >= 70:
    print("A ile geçtiniz.")
elif note >= 50:
    print("B ile geçtiniz.")
else:
    print("Kaldınız.")
```

## Nested if

```python
note = input("Lütfen notunuzu giriniz: ")
if note.isnumeric():
    note = int(note)
    if note >= 70:
        print("A ile geçtiniz.")
    elif note >= 50:
        print("B ile geçtiniz.")
    else:
        print("Kaldınız.")
else:
    print("Lütfen sayısal bir değer giriniz.")
```

# Ödev: Hava Durumu ve Kıyafet Önerileri

Kullanıcıdan **hava sıcaklığını** isteyin ve sıcaklığa göre uygun **kıyafet önerileri** verin.

Öneriler aşağıdaki gibi olabilir:

* **Sıcaklık 30°C ve üzerindeyse:**
  `"Bugün hava çok sıcak, hafif kıyafetler giyin."`

* **Sıcaklık 20°C ile 29°C arasındaysa:**
  `"Hava güzel, rahat kıyafetler tercih edebilirsiniz."`

* **Sıcaklık 10°C ile 19°C arasındaysa:**
  `"Hava biraz serin, hafif bir ceket giymeyi unutmayın."`

* **Sıcaklık 0°C ile 9°C arasındaysa:**
  `"Hava soğuk, sıcak tutan kıyafetler giymelisiniz."`

* **Sıcaklık 0°C'nin altındaysa:**
  `"Hava çok soğuk! Sıkı giyinin, hasta olmayın!"`
