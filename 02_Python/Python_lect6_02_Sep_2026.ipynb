## update()

```python
friends = { 
    "friend1": "Alice",
    "friend2": "Bob",
    "friend3": "Charlie"
}
friends
```

```python
friends.update({"friend4": "David", "friend5": "Eve"}) # update() metodu ile dict verisine yeni eleman eklenebilir.
friends
```

```python
friends.update(["feiend6", "Ann"]) # update içerisine dict veri tipinde eleman verilir.
```

## Nested Dictionaries

```python
my_family = {
    "wife": {"name": "Ayşe", "age": 30},
    "husband": {"name": "Mehmet", "age": 35},
    "children": {"son": {"name": "Ali", "age": 5}, 
                "daughter": {"name": "Elif", "age": 3}},
    "pets": {"dog": {"name": "Karabas", "age": 2}
}
}
my_family
```

```python
my_family["children"]["daughter"]["age"]
```

```python
my_family.keys()
```

```python
my_family["children"].keys()
```

```python
my_family.get("wife") # get() metodu ile dict verisinden key kullanarak value değerine ulaşılabilir.
```
