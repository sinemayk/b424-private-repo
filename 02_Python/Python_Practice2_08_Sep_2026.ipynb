## Egzersiz 1 — Dik Üçgen

İki dik kenarı verilen bir dik üçgenin **hipotenüsünü** bulmak için bir fonksiyon oluşturun.

Daha sonra bu üçgenin **çevresini** hesaplayın.

> **İpucu:** Hipotenüsü bulmak için Pisagor Teoremi'nden yararlanabilirsiniz.

````python
def hipotenus(x, y):
    hipo = (x ** 2 + y ** 2) ** 1/2
    cevre = x + y + hipo
    return f"kenar 1: {x}, kenar 2: {y}, hipotenüs: {hipo}, cevre: {cevre}"
````

````python
hipotenus(3, 4)
````

## Egzersiz 2 — İsimdeki Sesli Harfleri Sayma

Kullanıcıdan adını alın ve adında kaç tane **sesli harf** bulunduğunu hesaplayın.

Her bir sesli harf için **10 puan** verin.

Programın sonunda kullanıcının **kaç puan kazandığını** ekrana yazdırın.

````python
def sesli_puan(isim):
    sesli = "aeıioöuü"
    puan = 0
    isim = isim.lower()
    for harf in isim:
        if harf in sesli:
            puan += 10
    return puan


sesli_puan("Sinem")
````

## Egzersiz 3 — Asal Sayı ve Lasa Sayısı

Bir sayının **asal sayı** olup olmadığını kontrol eden bir fonksiyon oluşturun.

Daha sonra, sayının **lasa sayısı** olup olmadığını kontrol eden başka bir fonksiyon oluşturun.

> **Not:** Lasa sayı, tersten yazıldığında da asal olan sayıdır. Örneğin, `13` asal bir sayıdır ve tersi olan `31` de asal olduğu için `13` bir lasa sayıdır.

````python
def asal(sayi):
    i = 2
    z = True
    while z and i < sayi:
        if sayi % i == 0:
            z = False
        i += 1
    return z

def lasa(sayi):
    ters_sayi = int(str(sayi)[::-1])
    return asal(sayi) and asal(ters_sayi)
````

````python
lasa(11)
````

````python
lasa(8)
````

## Egzersiz 4 — Belirli Harfle Başlayan İsimler

Birden fazla isim alan bir fonksiyon oluşturun.

Fonksiyon, **`a` harfi ile başlayan isimleri** geri döndürsün.

* `*args` kullanın.
* Sonucu oluşturmak için **list comprehension** kullanın.

````python
def names(*args):
    return [i for i in args if i[0].lower() == "a"]
````

````python
names("Ayşe", "Berna", "Canan", "Derya", "Ahsen")
````

````python
isimler = ["Ayşe", "Berna", "Canan", "Derya", "Ahsen"]
````

````python
names(isimler) #args tuple olarak argüman aldığı için listeyi okumadı.
````

````python
names(*isimler)
````

````python
def ogrenci_bilgileri(**kwargs):  # Fonksiyona kaç tane bilgi gönderileceği bilinmiyorsa kwargs kullanılır.
    print(kwargs)
````

````python
ogrenci_bilgileri(ad = "Ali", yas = 22, bolum = "Biology")
````

## Egzersiz 5 — Liste Elemanlarının Ortalaması

Verilen bir sayı listesindeki sayıların **ortalamasını hesaplayan bir fonksiyon** oluşturun.

````python
def ortalama(*args):
    return sum(args) / len(args)
````

````python
sum((2,3,4))
````

````python
ortalama(1, 2, 3, 4, 5)
````

````python
def ort(list):
    toplam = 0
    for i in list:
        toplam += i
    return round(toplam / len(list))
````

````python
ort([1, 2, 3, 4, 5])
````


## Egzersiz 6 — İndirimli Ürün Fiyatları

Bir mağaza bazı ürünlerde indirim yapacaktır.

Verilen **orijinal fiyat listesindeki** ürünlerin indirimli fiyatlarını hesaplayan bir fonksiyon oluşturun.

Kurallar:

* İndirim oranı **%10**'dur.
* **100 TL'nin altındaki ürünlere indirim uygulanmayacaktır.**
* Fonksiyon, indirimli fiyatlardan oluşan bir **liste döndürmelidir**.
* Sonucu oluşturmak için **list comprehension** kullanın.
* List comprehension içerisinde **if-else** yapısını kullanın.

````python
def indirim(fiyat):
    return [i if i < 100 else i * 0.9 for i in fiyat]
````

````python
fiyat = [20, 50, 120, 380, 200]
indirim(fiyat)
````

## Egzersiz 7 — Öğrenci Not Ortalaması

İç içe listeler (**nested lists**) kullanarak öğrencilerin not ortalamalarını hesaplayın.

Her öğrencinin notlarından oluşan bir liste bulunmaktadır.

Öğrencilerin:

* Not ortalamalarını hesaplayın.
* Ortalaması **60 ve üzeri olan öğrencileri başarılı** olarak belirleyin.
* Ortalaması **60'ın altında olan öğrencileri başarısız** olarak belirleyin.


````python

students_grades = {
    "Ali": [70, 85, 90],
    "Ayşe": [45, 60, 55],
    "Mehmet": [80, 95, 92],
    "Zeynep": [50, 40, 65]
}
````

````python
def durum(student_grades):
    for i in student_grades:
        if ort(student_grades[i]) >= 60:
            print(f"{i} başarılı oldu.")
        else:
            print(f"{i} kaldı.")
````

````python
durum(students_grades)
````


## Egzersiz 8 — Arkadaş Sayılar

İki sayıya **arkadaş sayılar (amicable numbers)** denir, eğer birinci sayının kendisi hariç pozitif bölenlerinin toplamı ikinci sayıya ve ikinci sayının kendisi hariç pozitif bölenlerinin toplamı birinci sayıya eşitse.

Örneğin, **220 ve 284** arkadaş sayılardır.

Çünkü:

* 220'nin kendisi hariç pozitif bölenlerinin toplamı **284**'tür.
* 284'ün kendisi hariç pozitif bölenlerinin toplamı **220**'dir.

### Görev

İki sayının arkadaş sayılar olup olmadığını kontrol eden bir fonksiyon oluşturun.

````python
def arkadas(sayi1, sayi2):
    toplam_bolen_1 = sum([i for i in range(1, sayi1) if sayi1 % i == 0])
    toplam_bolen_2 = sum([i for i in range(1, sayi2) if sayi2 % i == 0])
    if toplam_bolen_1 == sayi2 and toplam_bolen_2 == sayi1:
        return f"{sayi1} ve {sayi2} arkadaş sayılardır"
    else:
        return f"{sayi1} ve {sayi2} arkadaş sayılar değildir."
    
arkadas(220, 284)
````

````python
arkadas(360, 420)
````

## Egzersiz 9 — Harshad Sayıları (Niven Sayıları)

Bir sayı, **basamaklarının toplamına tam olarak bölünebiliyorsa** Harshad sayısı (Niven sayısı) olarak adlandırılır.

Örneğin **18** bir Harshad sayısıdır.

Çünkü:

```text
1 + 8 = 9
```

ve:

```text
18 / 9 = 2
```

yani 18, basamakları toplamı olan 9'a tam olarak bölünmektedir.

### Görev

Verilen bir sayının Harshad sayısı olup olmadığını kontrol eden bir fonksiyon oluşturun.

````python
def harshad_sayi(sayi):
    str_sayi = str(sayi)
    rakam_topl = 0
    for i in str_sayi:
        rakam_topl += int(i)
    return sayi % rakam_topl == 0

harshad_sayi(18)
````

````python
harshad_sayi(56)
````

````python
def harshad(sayi):
    str_sayi = str(sayi)
    rakam_topl = sum([int(i) for i in str_sayi])
    return sayi % rakam_topl == 0
````

````python
harshad(18)
````
