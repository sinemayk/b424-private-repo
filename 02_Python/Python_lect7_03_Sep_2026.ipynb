## Union, Intersection, Difference

````python
A = set("suzan")
B = set("yavuz")
````

````python
print(A, B)
````

````python
A.union(B) # A ve B kümelerinin birleşimi
````

````python
A + B 
````

````python
A | B # A ve B kümelerinin birleşimi. Union sembolü | ile gösterilir.
````

````python
A.intersection(B) # A ve B kümelerinin kesişimi
````

````python
A & B # A ve B kümelerinin kesişimi. Intersection sembolü & ile gösterilir.
````

````python
A.difference(B) # A kümesinin B'den farkı.
````

````python
A - B # A kümesinin B'den farkı. Difference sembolü - ile gösterilir.
````

````python
B - A
````

````python
(A - B) | (B - A)
````

````python
python_kursu = {"Ali", "Veli", "Ayşe", "Fatma", "Can", "Deniz", "Ece", "Fırat"}
aws_kursu = {"Ayşe", "Fatma", "Can", "Deniz", "Ece", "Fırat"}
qa_kursu = {"Ali", "Sinem", "Canan", "Can", "Zehra"}
````

````python
# AWS kursu Python'ın alt kursu mu?
aws_kursu.issubset(python_kursu) # AWS kursu Python kursunun alt kümesi mi?
````

````python
# Hem qa hem Python kursalırını aan öğrencileri yazdırın.
python_kursu.intersection(qa_kursu) # Python ve QA kurslarının kesişimi.
````

````python
python_kursu & qa_kursu
````

````python
# Üç kursa birden katılan öğrenciler kimler?
python_kursu & qa_kursu & aws_kursu
````

````python
# Sadece Python kursuna katılan öğrenciler kimler?
python_kursu - qa_kursu - aws_kursu
````

````python
python_kursu.difference(qa_kursu).difference(aws_kursu) # Sadece Python kursuna katılan öğrenciler.
````

````python
# En az iki kursa katılan öğrenciler kimler?
en_az_iki_kurs_alanlar = (python_kursu & qa_kursu) | (python_kursu & aws_kursu) | (qa_kursu & aws_kursu) | (python_kursu & qa_kursu & aws_kursu)
````

````python
en_az_iki_kurs_alanlar
````

````python
# En az bir kurs alanlar
en_az_bir_kurs_alanlar = python_kursu | qa_kursu | aws_kursu
en_az_bir_kurs_alanlar
````

# 🎬 Egzersiz: Sinema Filmleri Dictionary

## 📌 Senaryo

Bir sinema salonu, şu anda gösterimde olan filmlerin **türlerini** ve **seans saatlerini** takip etmek istiyor.

## 📝 Talimatlar

### 1. Dictionary oluşturun

Aşağıdaki gibi bir `film_listesi` dictionary'si oluşturun:

```python
film_listesi = {
    "Inception": {"tur": "Bilim Kurgu", "seans": "15:00"},
    "The Godfather": {"tur": "Dram", "seans": "18:00"},
    "Pulp Fiction": {"tur": "Suc", "seans": "21:00"}
}

2. Yeni film ekleyin

"Interstellar" adında yeni bir film ekleyin.

Türü: "Bilim Kurgu"
Seansı: "17:00"
3. Film seansını güncelleyin

"Inception" filminin seans saatini "16:00" olarak güncelleyin.

4. Film silin

"The Godfather" filmini dictionary'den silin.

5. Güncellenmiş listeyi yazdırın

Yaptığınız tüm değişikliklerden sonra film_listesi dictionary'sinin güncel halini ekrana yazdırın.

````python
film_listesi = {
    "Inception": {"tur": "Bilim Kurgu", "seans": "15:00"},
    "The Godfather": {"tur": "Dram", "seans": "18:00"},
    "Pulp Fiction": {"tur": "Suc", "seans": "21:00"},
    "Matrix": {"tur": "Bilim Kurgu", "seans": "12:00"},
    "Titanic": {"tur": "Dram", "seans": "20:00"}
}
````

````python
film_listesi["Interstellar"] = {"tur": "Bilim Kurgu", "seans": "17:00"} # dict veri yapısının içerisine yeni bir key-value çifti ekler.
````

````python
film_listesi
````

````python
film_listesi["Inception"]["seans"] = "16:00" # dict veri yapısının içerisindeki bir key-value çiftinin value değerini değiştirir.
film_listesi
````

````python
film_listesi.pop("The Godfather") # dict veri yapısının içerisindeki belirtilen key-value çiftini siler ve silinen value değerini döndürür.
````

````python
film_listesi
````
