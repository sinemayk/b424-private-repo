## Soru 11 — Tech, Pro, Techpro

Kullanıcıdan bir sayı alınız.

- Sayı **3'e bölünebiliyorsa** `Tech` yazdırınız.
- Sayı **5'e bölünebiliyorsa** `Pro` yazdırınız.
- Sayı **hem 3'e hem de 5'e bölünebiliyorsa** `Techpro` yazdırınız.
- Sayı bu koşulların hiçbirini sağlamıyorsa **sayının kendisini** yazdırınız.

Bu işlemi gerçekleştiren Python kodunu yazınız.

````python
sayi = input("Lütfen sayi girin: ")

if sayi.replace(".", "", 1).isnumeric():
    sayi = int(sayi)

    if sayi % 3 == 0 and sayi % 5 == 0:
        print("Techpro")
    elif sayi % 3 == 0:
        print("Tech")
    elif sayi % 5 == 0:
        print("Pro")
    else:
        print(sayi)
else:
    print("Lütfen geçerli bir sayı giriniz.")
````

**Örneğin:** Kullanıcı `5`, `7` ve `multiply` değerlerini girerse, program `5 × 7` işlemini gerçekleştirmeli ve sonuç olarak **35** yazdırmalıdır.

Kullanıcı geçerli bir işlem operatörü girmezse aşağıdaki mesajı yazdırınız:

**`Please choose one of the operators(add,subtract,divide,multiple)`**


````python
print("Hesap makinesi programina hos geldiniz!")
ilk_sayi = input("Lutfen ilk sayiyi girin: ")
ilk_sayi = ilk_sayi.replace(",", ".", 1)
ilk_sayi = float(ilk_sayi)
ikinci_sayi = input("Lutfen ikinci sayiyi girin: ")
ikinci_sayi = ikinci_sayi.replace(",", ".", 1)
ikinci_sayi = float(ikinci_sayi)
islemler = input("Lutfen yapmak istediginiz islemi secin (+, -, *, /): ")

if islemler not in ["+", "-", "*", "/"]:
    print("Gecersiz islem secimi!")
else:
    if islemler == "+":
        sonuc = ilk_sayi + ikinci_sayi
        print(f"Toplam: {sonuc}")
    elif islemler == "-":
        sonuc = ilk_sayi - ikinci_sayi
        print(f"Fark: {sonuc}")
    elif islemler == "*":
        sonuc = ilk_sayi * ikinci_sayi
        print(f"Carpim: {sonuc}")
    elif islemler == "/":
        if ikinci_sayi != 0:
            sonuc = ilk_sayi / ikinci_sayi
            print(f"Bölüm: {sonuc}")
        else:
            print("Hata: Sifira bölme hatası!")
````

````python
number1 = input('Please enter the first number:')
number2 = input('Please enter the second number:')
operator = input('Please choose one of the operators (+ , -, /, *):')
if number1.replace('.', '', 1).replace('-', '', 1).isnumeric and number2.replace('.', '', 1).replace('-', '', 1).isnumeric:
    number1 = float(number1)
    number2 = float(number2)
    if (operator == '+'):
        print(f'{number1} {operator} {number2} = ', round((number1 + number2), 2))
    elif operator == '-':
        print(f'{number1} {operator} {number2} = ', round((number1 - number2), 2))
    elif operator == '/':
        if number2 == 0:
            print('Numbers cant be divided to 0')
        else:
            print(f'{number1} {operator} {number2} = ', round((number1 / number2), 2))
    elif operator == '*':
        print(f'{number1} {operator} {number2} = ', round((number1 * number2), 2))
    else:
        print('You entered wrong operator')
else:
    print('The data you entered supposed to be in numeric format')
````
