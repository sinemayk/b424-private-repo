# Escape Sequences

```python
print('I'm a python teacher.')
```

```python
print("I'm a python teacher.")
```

```python
print('I\'m a python theacher.')
```

| Escape sequence | Anlamı            | Örnek                 | Çıktı                      |
| --------------- | ----------------- | --------------------- | -------------------------- |
| `\n`            | Yeni satır        | `"Hello\nWorld"`      | Hello<br>World             |
| `\t`            | Tab               | `"Name:\tCanan"`      | Name: Canan                |
| `\\`            | Backslash         | `"C:\\Users"`         | C:\Users                   |
| `\'`            | Tek tırnak        | `'It\'s Python'`      | It's Python                |
| `\"`            | Çift tırnak       | `"He said \"Hello\""` | He said "Hello"            |
| `\r`            | Satır başına dön  | `"Hello\rWorld"`      | World                      |
| `\b`            | Backspace         | `"Hell\bo"`           | Helo                       |
| `\f`            | Form feed         | `"Hello\fWorld"`      | Özel kontrol karakteri     |
| `\a`            | Bell / uyarı sesi | `"\a"`                | Sistem desteğine bağlı ses |
| `\0`            | Null karakter     | `"Hello\0World"`      | Null karakter içerir       |

```python
print(not 6)
```

```python
print(not 0)
```

### and
* Eğer bütün elemanlar True ise en son değeri yazdırır.
* Eğer bir tane bile False varsa gördüğü ilk False değerini yazdırır.
* **and** mükemmeliyetçidir hep False arar.

```python
print(5 and 7 and 9)
```

```python
print(5 and 0 and 9 and ())
```

```python
print(9 and "Antalya" and 0)
```

### or

* Eğer bütün elemanlar False ise en son gördüğü değeri yazdırır.
* Eğer bir tane bile True varsa gördüğü ilk True değerini yazdırır.
* **or** optimisttir, her zaman True arar.

```python
print(0 or [] or False)
```

```python
print(0 or 7 or False or "Python")
```

```python
print(1 or "Techpro" or 7.5)
```
