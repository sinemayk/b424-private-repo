1. **Indentation (Girintileme):** Kod bloklarında **4 boşluk (space)** kullanılır.
2. **Variables (Değişkenler):** `snake_case` kullanılır. → `student_name`
3. **Functions (Fonksiyonlar):** `snake_case` kullanılır. → `calculate_average()`
4. **Classes (Sınıflar):** `PascalCase` kullanılır. → `DataProcessor`
5. **Constants (Sabitler):** `UPPER_CASE` kullanılır. → `MAX_VALUE`
6. **Imports:** `import` ifadeleri dosyanın **en üstünde** bulunmalıdır.
7. **Operators:** Operatörlerin etrafında boşluk bırakılır. → `x = a + b`
8. **Commas:** Virgüllerden sonra **bir boşluk** bırakılır. → `[1, 2, 3]`
9. **Naming:** Değişken, fonksiyon ve class isimleri **anlamlı ve açıklayıcı** olmalıdır.
10. **Quotes:** String ifadelerde **tutarlı bir tırnak stili** kullanılmalıdır.
11. **Comments:** Gereksiz yorumlardan kaçınılmalı; yorumlar özellikle **neden** sorusunu açıklamalıdır.
12. **Line Length:** Kod satırları gereksiz şekilde uzun tutulmamalı, **okunabilirlik korunmalıdır**.
13. **Blank Lines:** Top-level function ve class tanımları arasında **2 boş satır** bırakılmalıdır.
14. **Whitespace:** Gereksiz boşluklardan kaçınılmalıdır. → `x = [1, 2, 3]`
15. **Consistency:** Aynı proje içerisinde **aynı kodlama stili tutarlı şekilde** uygulanmalıdır.
