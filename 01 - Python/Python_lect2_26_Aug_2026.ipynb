```python
# Kullanıcıdan aldığınız boy ve kilo bilgisi ile vücut kitle indeksini hesaplayınız.

boy = float(input("Boyunuzu metre cinsinden giriniz: "))
kilo = int(input("Kilonuzu kilogram cinsinden giriniz: "))

vki = kilo / (boy ** 2)

print(f"Boyunuz {boy} metre, kilonuz {kilo} kilogram olduğuna göre vücut kitle indeksiniz: {vki}'dir.")
```

```python
boy = float(input("Boyunuzu metre cinsinden giriniz: "))
kilo = int(input("Kilonuzu kilogram cinsinden giriniz: "))

vki = kilo / (boy ** 2)

print(f"Kullanıcın boyu {boy} metre, kilosu {kilo} kilogram olduğuna göre vücut kitle indeksi: {round(vki, 2)}'dir.")
```
